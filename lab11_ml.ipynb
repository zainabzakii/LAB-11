```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
cc_data = pd.read_csv('CC_GENERAL.csv')
```

```python
cc_data.head()
```

```python
cc_data.info()
```

```python
cc_data.describe()
```

```python
cc_data.isnull().sum()
```

```python
cc_data.fillna(cc_data.mean(numeric_only=True), inplace=True)
```

```python
cc_data.drop('CUST_ID', axis=1, inplace=True)
```

```python
cc_data[['BALANCE','PURCHASES','CREDIT_LIMIT']].hist(
    figsize=(12,4),
    bins=20
)

plt.tight_layout()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    cc_data['PURCHASES'],
    cc_data['CREDIT_LIMIT']
)

plt.xlabel('PURCHASES')
plt.ylabel('CREDIT_LIMIT')
plt.title('Customer Data Before Clustering')

plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(cc_data)
```

```python
from sklearn.cluster import KMeans

inertia = []
K = range(1,11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_data)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))
plt.plot(K, inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Choosing K')
plt.xticks(K)

plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []
K_range = range(2,11)
for k in K_range:

    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(scaled_data)
    silhouette_scores.append(
        silhouette_score(scaled_data, labels)
    )
```

```python
plt.figure(figsize=(8,5))
plt.plot(K_range, silhouette_scores, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score for Different K Values')
plt.xticks(K_range)

plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
cc_data['Cluster'] = kmeans.fit_predict(scaled_data)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    cc_data['PURCHASES'],
    cc_data['CREDIT_LIMIT'],
    c=cc_data['Cluster'],
    cmap='viridis'
)

plt.xlabel('PURCHASES')
plt.ylabel('CREDIT_LIMIT')
plt.title('Customer Segments using K-Means')

plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    cc_data['PURCHASES'],
    cc_data['CREDIT_LIMIT'],
    c=cc_data['Cluster'],
    cmap='viridis'
)

plt.scatter(
    cc_data.groupby('Cluster')['PURCHASES'].mean(),
    cc_data.groupby('Cluster')['CREDIT_LIMIT'].mean(),
    c='black',
    marker='X',
    s=200,
    label='Centroids'
)

plt.xlabel('PURCHASES')
plt.ylabel('CREDIT_LIMIT')
plt.title('Clusters with Centroids')
plt.legend()

plt.show()
```

```python
fig, axes = plt.subplots(2,3, figsize=(15,10))
axes = axes.flatten()
for i, k in enumerate(range(2,8)):

    kmeans = KMeans(n_clusters=k, random_state=42)
    clusters = kmeans.fit_predict(scaled_data)
    axes[i].scatter(
        cc_data['PURCHASES'],
        cc_data['CREDIT_LIMIT'],
        c=clusters,
        cmap='Dark2'
    )

    axes[i].set_title(f'K = {k}')
plt.tight_layout()
```

```python
final_kmeans = KMeans(n_clusters=4, random_state=42)
cc_data['Final Cluster'] = final_kmeans.fit_predict(scaled_data)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    cc_data['PURCHASES'],
    cc_data['CREDIT_LIMIT'],
    c=cc_data['Final Cluster'],
    cmap='viridis'
)

plt.scatter(
    cc_data.groupby('Final Cluster')['PURCHASES'].mean(),
    cc_data.groupby('Final Cluster')['CREDIT_LIMIT'].mean(),
    c='black',
    marker='X',
    s=200,
    label='Centroids'
)

plt.xlabel('PURCHASES')
plt.ylabel('CREDIT_LIMIT')
plt.title('Final Customer Segments')
plt.legend()

plt.show()
```
